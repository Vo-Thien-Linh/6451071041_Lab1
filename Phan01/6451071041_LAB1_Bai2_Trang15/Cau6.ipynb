```python
from pathlib import Path
import pandas as pd
from IPython.display import display

candidates = [Path('titanic.csv'), Path('LAB1') / 'titanic.csv', Path('LAB1') / 'Bai2_Trang15-17' / 'titanic.csv', Path('../titanic.csv'), Path('/content/titanic.csv'), Path('/mnt/data/titanic.csv')]
data_path = next((p for p in candidates if p.is_file()), None)
if data_path is None:
    try:
        from google.colab import files
    except ImportError:
        files = None
    if files is not None:
        print('Chưa tìm thấy titanic.csv. Hãy chọn tệp CSV để tải lên Colab.')
        uploaded = files.upload()
        uploaded_path = next((Path(name) for name in uploaded if Path(name).name.lower() == 'titanic.csv'), None)
        if uploaded_path is not None and uploaded_path.is_file():
            data_path = uploaded_path
if data_path is None:
    raise FileNotFoundError(f'Không tìm thấy titanic.csv. Thư mục chạy hiện tại: {Path.cwd()}. Hãy tải CSV lên Colab hoặc đặt CSV trong thư mục này rồi chạy lại ô.')
df = pd.read_csv(data_path)
print('Bảng 1. Số giá trị thiếu trước xử lý')
missing_before = (df.isna().sum().sort_values(ascending=False)
                 .rename_axis('Thuộc tính').reset_index(name='Số giá trị thiếu'))
display(missing_before)
df['age'] = df['age'].fillna(df['age'].median())
df['fare'] = df['fare'].fillna(df['fare'].median())
if df['embarked'].isna().any():
    df['embarked'] = df['embarked'].fillna(df['embarked'].mode().iloc[0])
drop_columns = ['cabin', 'boat', 'body', 'ticket', 'home.dest']
df = df.drop(columns=drop_columns, errors='ignore')
print('\nBảng 2. Số giá trị thiếu sau xử lý')
missing_after = (df.isna().sum().sort_values(ascending=False)
                .rename_axis('Thuộc tính').reset_index(name='Số giá trị thiếu'))
display(missing_after)
print(f'\nKích thước sau xử lý: {df.shape[0]} dòng × {df.shape[1]} cột')
print('Các cột còn lại:', df.columns.tolist())
print('\nBảng 3A. Các thuộc tính hành khách (5 dòng đầu; tiêu đề giữ nguyên tên cột dữ liệu)')
display(df[['pclass', 'survived', 'name', 'sex', 'age']].head())
print('Bảng 3B. Người thân đi cùng, giá vé và cảng lên tàu (5 dòng đầu)')
display(df[['sibsp', 'parch', 'fare', 'embarked']].head())
```
