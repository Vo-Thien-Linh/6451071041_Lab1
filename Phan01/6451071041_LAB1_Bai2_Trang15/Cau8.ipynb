# Câu 8 — Tạo thuộc tính mới

**Sinh viên: Võ Thiện Linh**

Tạo `age_class = age × pclass` và `fare_per_person = fare / (sibsp + parch + 1)`. Số 1 ở mẫu số tính chính hành khách, nên tránh chia cho 0 khi hành khách đi một mình. Notebook tự xử lý các giá trị thiếu cần thiết và chạy độc lập.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

candidates = [Path('titanic.csv'), Path('LAB1') / 'titanic.csv', Path('LAB1') / 'Bai2_Trang15-17' / 'titanic.csv', Path('../titanic.csv'), Path('/content/titanic.csv'), Path('/mnt/data/titanic.csv')]
data_path = next((p for p in candidates if p.is_file()), None)
if data_path is None:
    try:
        from google.colab import files
    except ImportError:
        files = None
    if files is not None:
        print('Chưa tìm thấy titanic.csv. Hãy chọn tệp CSV để tải lên Colab.')
        uploaded = files.upload()
        uploaded_path = next((Path(name) for name in uploaded if Path(name).name.lower() == 'titanic.csv'), None)
        if uploaded_path is not None and uploaded_path.is_file():
            data_path = uploaded_path
if data_path is None:
    raise FileNotFoundError(f'Không tìm thấy titanic.csv. Thư mục chạy hiện tại: {Path.cwd()}. Hãy tải CSV lên Colab hoặc đặt CSV trong thư mục này rồi chạy lại ô.')
df = pd.read_csv(data_path)
df['age'] = df['age'].fillna(df['age'].median())
df['fare'] = df['fare'].fillna(df['fare'].median())
df['age_class'] = df['age'] * df['pclass']
df['fare_per_person'] = df['fare'] / (df['sibsp'] + df['parch'] + 1)
print('Đã tạo age_class = age × pclass và fare_per_person = fare / (sibsp + parch + 1).')
print('Các thuộc tính tuổi và hạng vé (10 dòng đầu):')
display(df[['age', 'pclass', 'age_class']].head(10))
print('Giá vé, số người trong gia đình và giá vé mỗi người (10 dòng đầu):')
display(df[['fare', 'sibsp', 'parch', 'fare_per_person']].head(10))
print('\nThống kê hai thuộc tính mới:')
display(df[['age_class', 'fare_per_person']].describe().T.rename_axis('attribute').reset_index())
```
