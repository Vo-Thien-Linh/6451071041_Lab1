# Câu 7 — Biến đổi thuộc tính Bank Marketing

**Võ Thiện Linh — MSSV: 6451071041**

Mã hóa `default`, `loan`, `subscribe`; One-hot Encode các thuộc tính phân loại; chia nhóm `duration` và `cons_price_idx`. Chỉ số giá có ít giá trị duy nhất nên số nhóm phân vị có thể ít hơn 6 để không tách các giá trị bằng nhau.

```python
print('Võ Thiện Linh - MSSV: 6451071041')
import pandas as pd
from IPython.display import display
rows = [['housemaid','Married','basic.4y','no','no',261,93.994,'no'],['services','Married','high.school','unknown','no',149,93.994,'no'],['blue-collar','Divorced','basic.4y','unknown','no',1575,93.994,'yes'],['admin.','Single','high.school','no','no',338,93.994,'no'],['blue-collar','Single','basic.9y','no','no',179,93.994,'no'],['services','Married','high.school','no','no',1030,93.994,'yes'],['management','Married','high.school','unknown','no',149,None,'no'],['unemployed',None,'university.degree','unknown','no',424,93.994,'no'],['technician','Married','professional.course','no','no',1623,93.994,'yes'],['services','Divorced','high.school','unknown','no',568,93.994,'no'],['blue-collar','Married','high.school','no','no',1297,93.994,'yes'],['self-employed','Married','basic.9y','no','no',376,93.918,'no'],['entrepreneur','Married','professional.course','no','yes',576,93.994,'no'],['services','Single','high.school','no','no',1059,93.2,'yes'],['technician','Married','basic.9y','no','no',705,93.2,'yes'],['retired','Married','high.school','no','no',532,93.2,'no']]
df = pd.DataFrame(rows,columns=['job','marital','education','default','loan','duration','cons_price_idx','subscribe'])
df['marital'] = df['marital'].fillna('unknown')
df['cons_price_idx'] = df['cons_price_idx'].fillna(df['cons_price_idx'].median())
df['duration'] = pd.to_numeric(df['duration'],errors='raise').astype('int64')
df['default_code'] = df['default'].map({'no':0,'yes':1,'unknown':-1}).astype('int64')
df['loan_code'] = df['loan'].map({'no':0,'yes':1}).astype('int64')
df['subscribe_code'] = df['subscribe'].map({'no':0,'yes':1}).astype('int64')
df['duration_group'] = pd.qcut(df['duration'].rank(method='first'),q=8,labels=False)+1
df['cons_price_idx_group'] = pd.qcut(df['cons_price_idx'],q=6,labels=False,duplicates='drop')+1
print('Các thuộc tính đã chuyển đổi:')
display(df[['duration','default','default_code','loan','loan_code','subscribe','subscribe_code','duration_group','cons_price_idx','cons_price_idx_group']].head(10))
print('Số dòng trong mỗi nhóm duration:')
display(df['duration_group'].value_counts().sort_index().rename_axis('duration_group').reset_index(name='row_count'))
print('Số dòng trong nhóm cons_price_idx:')
display(df.groupby('cons_price_idx_group',observed=True)['cons_price_idx'].agg(row_count='size',min_value='min',max_value='max').reset_index())
one_hot = pd.get_dummies(df[['job','marital','education']],dtype=int)
print('One-hot Encoding:')
display(one_hot.head(10))
print('duration giữ kiểu nguyên:',df['duration'].dtype)
```
