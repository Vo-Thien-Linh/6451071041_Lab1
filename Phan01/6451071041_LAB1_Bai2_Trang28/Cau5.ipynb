Thống kê số khách hàng và tỷ lệ `subscribe` theo nhóm `duration`.

```python
print('Võ Thiện Linh - MSSV: 6451071041')
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
rows = [['housemaid','Married','basic.4y','no','no',261,93.994,'no'],['services','Married','high.school','unknown','no',149,93.994,'no'],['blue-collar','Divorced','basic.4y','unknown','no',1575,93.994,'yes'],['admin.','Single','high.school','no','no',338,93.994,'no'],['blue-collar','Single','basic.9y','no','no',179,93.994,'no'],['services','Married','high.school','no','no',1030,93.994,'yes'],['management','Married','high.school','unknown','no',149,None,'no'],['unemployed',None,'university.degree','unknown','no',424,93.994,'no'],['technician','Married','professional.course','no','no',1623,93.994,'yes'],['services','Divorced','high.school','unknown','no',568,93.994,'no'],['blue-collar','Married','high.school','no','no',1297,93.994,'yes'],['self-employed','Married','basic.9y','no','no',376,93.918,'no'],['entrepreneur','Married','professional.course','no','yes',576,93.994,'no'],['services','Single','high.school','no','no',1059,93.2,'yes'],['technician','Married','basic.9y','no','no',705,93.2,'yes'],['retired','Married','high.school','no','no',532,93.2,'no']]
df = pd.DataFrame(rows,columns=['job','marital','education','default','loan','duration','cons_price_idx','subscribe'])
df['duration_group'] = pd.qcut(df['duration'].rank(method='first'),q=6,labels=False)+1
summary = df.groupby('duration_group').agg(duration_min=('duration','min'),duration_max=('duration','max'),customer_count=('subscribe','size'),subscribe_count=('subscribe',lambda s:s.eq('yes').sum()),subscribe_rate=('subscribe',lambda s:s.eq('yes').mean())).reset_index()
summary['subscribe_rate_percent'] = (summary['subscribe_rate']*100).round(1)
display(summary.drop(columns='subscribe_rate'))
ax = summary.plot(kind='bar',x='duration_group',y='subscribe_rate_percent',legend=False,color='#f2a541',figsize=(8,4))
ax.set_title('Tỷ lệ đăng ký theo nhóm duration')
ax.set_xlabel('duration_group (chia theo phân vị)')
ax.set_ylabel('subscribe rate (%)')
ax.tick_params(axis='x',rotation=0)
ax.grid(axis='y',alpha=0.25)
plt.tight_layout()
plt.show()
print('Nhận xét: đối chiếu tỷ lệ và số khách mỗi nhóm; mẫu 16 dòng chỉ mang tính minh họa.')
```
