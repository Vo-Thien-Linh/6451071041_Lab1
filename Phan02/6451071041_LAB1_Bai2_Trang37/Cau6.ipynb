# Online Retail — Câu 6: Dạng hot encoding theo hóa đơn

**Võ Thiện Linh — MSSV: 6451071041**

Từ tổng `Quantity` theo hóa đơn–mặt hàng, xoay dữ liệu thành ma trận số lượng: mỗi dòng là một hóa đơn, mỗi cột là một mặt hàng. Ma trận dùng cột thưa để tiết kiệm bộ nhớ.

```python
print('Võ Thiện Linh - MSSV: 6451071041')
from pathlib import Path
import numpy as np
import pandas as pd
from IPython.display import display
candidates = [Path('Online Retail.xlsx'), Path('LAB1') / 'Phan02' / '6451071041_LAB1_Bai2_Trang37' / 'Online Retail.xlsx', Path('/content/Online Retail.xlsx')]
data_path = next((p for p in candidates if p.is_file()), None)
if data_path is None:
    raise FileNotFoundError(f'Không tìm thấy Online Retail.xlsx. Thư mục đang chạy: {Path.cwd()}')
df = pd.read_excel(data_path, engine='openpyxl')
df['Description'] = df['Description'].str.strip()
df = df.dropna(subset=['InvoiceNo']).copy()
df['InvoiceNo'] = df['InvoiceNo'].astype('string').str.strip()
df = df.loc[~df['InvoiceNo'].str.contains('C', regex=False, na=False)]
uk = df.loc[df['Country'].eq('United Kingdom')]
grouped_quantity = uk.groupby(['InvoiceNo', 'Description'], as_index=False, dropna=True)['Quantity'].sum()
invoice_index = pd.Index(grouped_quantity['InvoiceNo'].drop_duplicates(), name='InvoiceNo')
row_position = {invoice: i for i, invoice in enumerate(invoice_index)}
sparse_columns = {}
for description, rows in grouped_quantity.groupby('Description', sort=False):
    values = np.zeros(len(invoice_index), dtype=np.int32)
    positions = rows['InvoiceNo'].map(row_position).to_numpy(dtype=np.intp)
    values[positions] = rows['Quantity'].to_numpy(dtype=np.int32)
    sparse_columns[description] = pd.arrays.SparseArray(values, fill_value=0)
basket_quantity = pd.DataFrame(sparse_columns, index=invoice_index)
print(f'Hot-encoding matrix: {basket_quantity.shape[0]:,} hóa đơn × {basket_quantity.shape[1]:,} mặt hàng')
print(f'Bộ nhớ ma trận thưa: {basket_quantity.memory_usage(index=True, deep=True).sum() / 1024**2:.2f} MiB')
print('Ma trận số lượng — 5 hóa đơn đầu, 10 mặt hàng đầu:')
display(basket_quantity.iloc[:5, :10].sparse.to_dense())
```
