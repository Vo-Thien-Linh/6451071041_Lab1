Áp dụng tiền xử lý ở Câu 2, sau đó xóa các hóa đơn có ký tự `C` trong `InvoiceNo`.

```python
print('Võ Thiện Linh - MSSV: 6451071041')
from pathlib import Path
import pandas as pd
from IPython.display import display
candidates = [Path('Online Retail.xlsx'), Path('LAB1') / 'Phan02' / '6451071041_LAB1_Bai2_Trang37' / 'Online Retail.xlsx', Path('/content/Online Retail.xlsx')]
data_path = next((p for p in candidates if p.is_file()), None)
if data_path is None:
    raise FileNotFoundError(f'Không tìm thấy Online Retail.xlsx. Thư mục đang chạy: {Path.cwd()}')
df = pd.read_excel(data_path, engine='openpyxl')
df['Description'] = df['Description'].str.strip()
df = df.dropna(subset=['InvoiceNo']).copy()
df['InvoiceNo'] = df['InvoiceNo'].astype('string').str.strip()
credit_mask = df['InvoiceNo'].str.contains('C', regex=False, na=False)
credit_rows = int(credit_mask.sum())
credit_invoices = int(df.loc[credit_mask, 'InvoiceNo'].nunique())
df = df.loc[~credit_mask].copy()
assert not df['InvoiceNo'].str.contains('C', regex=False, na=False).any()
print(f'Dòng thuộc hóa đơn tín dụng đã xóa: {credit_rows:,}')
print(f'Số hóa đơn tín dụng: {credit_invoices:,}')
print(f'Dòng còn lại: {len(df):,}')
print('Năm dòng sau khi xóa hóa đơn tín dụng:')
display(df.head())
```
