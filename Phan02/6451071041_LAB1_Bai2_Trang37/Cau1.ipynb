Đọc workbook `Online Retail.xlsx`, kiểm tra kích thước, tên cột và xem trước dữ liệu.

```python
print('Võ Thiện Linh - MSSV: 6451071041')
from pathlib import Path
import pandas as pd
from IPython.display import display
candidates = [Path('Online Retail.xlsx'), Path('LAB1') / 'Phan02' / '6451071041_LAB1_Bai2_Trang37' / 'Online Retail.xlsx', Path('/content/Online Retail.xlsx')]
data_path = next((p for p in candidates if p.is_file()), None)
if data_path is None:
    raise FileNotFoundError(f'Không tìm thấy Online Retail.xlsx. Thư mục đang chạy: {Path.cwd()}')
df = pd.read_excel(data_path, engine='openpyxl')
print(f'Đã đọc: {data_path.resolve()}')
print(f'Kích thước: {df.shape[0]:,} dòng × {df.shape[1]} cột')
print('Tên cột:', df.columns.tolist())
print('Kiểu dữ liệu:')
display(df.dtypes.rename_axis('column').reset_index(name='dtype'))
print('Năm dòng đầu:')
display(df.head())
```
